Prompt Solto vs. Prompt Testável

Este notebook demonstra a diferença prática na qualidade das respostas de um modelo de linguagem (LLM).

```python
!pip install -q -U openai

from openai import OpenAI

API_KEY = ""

client = OpenAI(api_key=API_KEY)
MODELO = "gpt-4o-mini"

print("Ambiente OpenAI configurado com sucesso!")
```

#Exemplo 1



```python
prompt_solto = f"Crie um post sobre IA para o LinkedIn"

response_solto = client.chat.completions.create(
    model=MODELO,
    messages=[
        {"role": "user", "content": prompt_solto}
    ]
)

print("--- RESULTADO DO PROMPT SOLTO ---")
print(response_solto.choices[0].message.content)
```

```python
prompt_melhor = f"Crie um post sobre IA para o LinkedIn, Escreva para gestores de produto. Defenda 1 ideia, use exemplo concreto, limite a 900 caracteres, termine com pergunta e evite hype."

response_melhor = client.chat.completions.create(
    model=MODELO,
    messages=[
        {"role": "user", "content": prompt_melhor}
    ]
)

print("--- RESULTADO DO PROMPT melhor ---")
print(response_solto.choices[0].message.content)
```
